```python
%sql
CREATE OR REPLACE TABLE global_enterprises.silver.transactions
USING DELTA
AS
SELECT o.id, o.quantity, o.order_date, p.unit_price, p.product_name, o.quantity * p.unit_price AS revenue, c.name, c.email
FROM global_enterprises.bronze.orders AS o
JOIN global_enterprises.bronze.products AS p
ON o.product_id = p.id
JOIN global_enterprises.bronze.customers AS c
ON o.customer_id = c.id;
```
