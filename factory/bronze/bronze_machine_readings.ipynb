```python
machine_readings = spark.read.csv("/Workspace/Users/rafaelngoncalves5@outlook.com/ge_databricks/factory/raw/machine_readings.csv", inferSchema="true")

machine_readings = machine_readings.withColumnRenamed("_c0", "reading_id")
machine_readings = machine_readings.withColumnRenamed("_c1", "machine_id")
machine_readings = machine_readings.withColumnRenamed("_c2", "timestamp")
machine_readings = machine_readings.withColumnRenamed("_c3", "temperature")
machine_readings = machine_readings.withColumnRenamed("_c4", "vibration")
machine_readings = machine_readings.withColumnRenamed("_c5", "pressure")
machine_readings = machine_readings.withColumnRenamed("_c6", "status")

display(machine_readings)
```

```python
# Drop duplicates
machine_readings = machine_readings.dropDuplicates()
```

```python
from pyspark.sql.functions import col, expr

# timestamp column
machine_readings = machine_readings.fillna({"timestamp": "1900-01-01 00:00:00"})
machine_readings = machine_readings.withColumn("timestamp", expr("try_cast(timestamp as timestamp)"))

# temperature column
machine_readings = machine_readings.withColumn("temperature", col("temperature").cast("double"))
machine_readings = machine_readings.fillna({"temperature": 0})
display(machine_readings)
```
