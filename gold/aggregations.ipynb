# Loading silver

```python
# Displaying the tables under default schema
spark.sql('SHOW CATALOGS').show()

spark.sql('SHOW SCHEMAS IN global_enterprises').show()

spark.sql('SHOW TABLES IN global_enterprises.silver').show()
```

```python
transactions = spark.read.table("global_enterprises.silver.transactions")
display(transactions)
```

# Curation

```python
from pyspark.sql import functions as F

# Revenue by product
revenue_by_product = transactions.groupBy("product_name").agg(F.sum("revenue").alias("revenue"))

# Revenue by customer
revenue_by_customer = transactions.groupBy("name").agg(F.sum("revenue").alias("revenue"))

# Saving the tables
revenue_by_product.write.mode("overwrite").saveAsTable("global_enterprises.gold.revenue_by_product")
revenue_by_customer.write.mode("overwrite").saveAsTable("global_enterprises.gold.revenue_by_customer")
```

```python
spark.read.table("global_enterprises.gold.revenue_by_product").show()
spark.read.table("global_enterprises.gold.revenue_by_customer").show()
```
