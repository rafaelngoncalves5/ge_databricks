```python
transactions = spark.read.table("global_enterprises.silver.transactions")
display(transactions)
```

# Curation

```python
%sql

DROP TABLE IF EXISTS global_enterprises.gold.revenue_by_month;
```

```python
from pyspark.sql import functions as F

# Revenue by product
revenue_by_product = transactions.groupBy("product_name").agg(F.sum("revenue").alias("revenue"))

# Revenue by customer
revenue_by_customer = transactions.groupBy("name").agg(F.sum("revenue").alias("revenue"))

# Revenue by Month
revenue_by_month = transactions.groupBy(F.date_format("order_date", "yyyy-MM").alias("month")).agg(F.sum("revenue").alias("revenue"))

# Revenue by month and product
revenue_by_month_product = transactions.groupBy(F.date_format("order_date", "yyyy-MM").alias("month"), "product_name").agg(F.sum("revenue").alias("revenue"))

# Revenue by month and customer
revenue_by_month_customer = transactions.groupBy(F.date_format("order_date", "yyyy-MM").alias("month"), "name").agg(F.sum('revenue').alias("revenue"))
```

```python
# Saving the tables
revenue_by_product.write.mode("overwrite").saveAsTable("global_enterprises.gold.revenue_by_product")
revenue_by_customer.write.mode("overwrite").saveAsTable("global_enterprises.gold.revenue_by_customer")
revenue_by_month.write.mode("overwrite").saveAsTable("global_enterprises.gold.revenue_by_month")
revenue_by_month_customer.write.mode("overwrite").saveAsTable("global_enterprises.gold.revenue_by_month_customer")
revenue_by_month_product.write.mode("overwrite").saveAsTable("global_enterprises.gold.revenue_by_month_product")
```
