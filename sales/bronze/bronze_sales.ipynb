```python
df_customers = spark.read.option('inferSchema', 'true').csv('/Volumes/global_enterprises/ingestion/raw/customers.csv')

df_orders = spark.read.option('inferSchema', 'true').csv('/Volumes/global_enterprises/ingestion/raw/orders.csv')

df_products = spark.read.option('inferSchema', 'true').csv('/Volumes/global_enterprises/ingestion/raw/products.csv')

# Adding columns
df_customers = df_customers.withColumnRenamed("_c0", "id")
df_customers = df_customers.withColumnRenamed("_c1", "name")
df_customers = df_customers.withColumnRenamed("_c2", "email")

df_orders = df_orders.withColumnRenamed("_c0", "id")
df_orders = df_orders.withColumnRenamed("_c1", "quantity")
df_orders = df_orders.withColumnRenamed("_c2", "order_date")
df_orders = df_orders.withColumnRenamed("_c3", "product_id")
df_orders = df_orders.withColumnRenamed("_c4", "customer_id")

df_products = df_products.withColumnRenamed("_c0", "id")
df_products = df_products.withColumnRenamed("_c1", "unit_price")
df_products = df_products.withColumnRenamed("_c2", "product_name")

# Check
display(df_customers)
display(df_orders)
display(df_products)
```

```python
# Drop tables first if they exist without schema
spark.sql('DROP TABLE IF EXISTS global_enterprises.bronze.customers')
spark.sql('DROP TABLE IF EXISTS global_enterprises.bronze.orders')
spark.sql('DROP TABLE IF EXISTS global_enterprises.bronze.products')

df_customers.write.mode('overwrite').saveAsTable('global_enterprises.bronze.customers')
df_orders.write.mode('overwrite').saveAsTable('global_enterprises.bronze.orders')
df_products.write.mode('overwrite').saveAsTable('global_enterprises.bronze.products')
```
